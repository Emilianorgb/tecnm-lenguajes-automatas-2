````python
%%writefile api_sesion10.h
/* api_sesion10.h - contrato de la Sesion 10
 * Mini-evaluador de expresiones (Hito 1)
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION10_H
#define API_SESION10_H

long evaluar_expresion(const char *fuente, int *error);

#endif /* API_SESION10_H */
````

````python
%%writefile artesano_sesion10_c.c
/* artesano_sesion10_c.c - Fase 1: El Artesano (sin IA)
 * Mini-evaluador de expresiones (Hito 1).
 *
 * Implementa lo que declara api_sesion10.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion10.h"
#include <stdio.h>
#include <string.h>
#include <ctype.h>

typedef struct {
    const char *fuente;
    int posicion;
    int error;
} Parser;

static void ignorar_espacios(Parser *p) {
    while (isspace((unsigned char)p->fuente[p->posicion])) {
        p->posicion++;
    }
}

static long analizar_expresion(Parser *p);

static long analizar_factor(Parser *p) {
    long valor = 0;
    int signo = 1;

    ignorar_espacios(p);

    if (p->fuente[p->posicion] == '+') {
        p->posicion++;
    } else if (p->fuente[p->posicion] == '-') {
        signo = -1;
        p->posicion++;
    }

    ignorar_espacios(p);

    if (p->fuente[p->posicion] == '(') {
        p->posicion++;

        valor = analizar_expresion(p);

        ignorar_espacios(p);

        if (p->fuente[p->posicion] != ')') {
            p->error = 2;
            return 0;
        }

        p->posicion++;
        return signo * valor;
    }

    if (!isdigit((unsigned char)p->fuente[p->posicion])) {
        p->error = 2;
        return 0;
    }

    while (isdigit((unsigned char)p->fuente[p->posicion])) {
        valor = valor * 10 + (p->fuente[p->posicion] - '0');
        p->posicion++;
    }

    return signo * valor;
}

static long analizar_termino(Parser *p) {
    long resultado = analizar_factor(p);

    while (p->error == 0) {
        char operador;

        ignorar_espacios(p);
        operador = p->fuente[p->posicion];

        if (operador != '*' && operador != '/') {
            break;
        }

        p->posicion++;

        long segundo = analizar_factor(p);

        if (p->error != 0) {
            return 0;
        }

        if (operador == '*') {
            resultado *= segundo;
        } else {
            if (segundo == 0) {
                p->error = 1;
                return 0;
            }

            resultado /= segundo;
        }
    }

    return resultado;
}

static long analizar_expresion(Parser *p) {
    long resultado = analizar_termino(p);

    while (p->error == 0) {
        char operador;

        ignorar_espacios(p);
        operador = p->fuente[p->posicion];

        if (operador != '+' && operador != '-') {
            break;
        }

        p->posicion++;

        long segundo = analizar_termino(p);

        if (p->error != 0) {
            return 0;
        }

        if (operador == '+') {
            resultado += segundo;
        } else {
            resultado -= segundo;
        }
    }

    return resultado;
}

long evaluar_expresion(const char *fuente, int *error) {
    Parser parser;
    long resultado;

    if (error == NULL) {
        return 0;
    }

    *error = 0;

    if (fuente == NULL) {
        *error = 2;
        return 0;
    }

    parser.fuente = fuente;
    parser.posicion = 0;
    parser.error = 0;

    resultado = analizar_expresion(&parser);

    ignorar_espacios(&parser);

    if (parser.error == 0 && parser.fuente[parser.posicion] != '\0') {
        parser.error = 2;
    }

    *error = parser.error;

    if (*error != 0) {
        return 0;
    }

    return resultado;
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle
la misma función a la IA, y pega el resultado en la celda de `ia_sesion10_c.c`. La
firma debe ser idéntica a la del contrato; si la IA la cambia, corrígela antes de
compilar.


````python
%%writefile ia_sesion10_c.c
/* ia_sesion10_c.c - Fase 2: El Cientifico
 *
 * Pega aqui el codigo que te devolvio la IA, con la MISMA firma
 * que declara api_sesion10.h. No puede traer main().
 */
#include "api_sesion10.h"
#include <ctype.h>
#include <stddef.h>

typedef struct {
    const char *texto;
    int indice;
    int codigo_error;
} Evaluador;

static void saltar_espacios(Evaluador *e) {
    while (isspace((unsigned char)e->texto[e->indice])) {
        e->indice++;
    }
}

static long expresion(Evaluador *e);

static long numero_o_parentesis(Evaluador *e) {
    long numero = 0;
    int negativo = 0;

    saltar_espacios(e);

    if (e->texto[e->indice] == '-') {
        negativo = 1;
        e->indice++;
    } else if (e->texto[e->indice] == '+') {
        e->indice++;
    }

    saltar_espacios(e);

    if (e->texto[e->indice] == '(') {
        e->indice++;

        numero = expresion(e);

        saltar_espacios(e);

        if (e->texto[e->indice] != ')') {
            e->codigo_error = 2;
            return 0;
        }

        e->indice++;

        return negativo ? -numero : numero;
    }

    if (!isdigit((unsigned char)e->texto[e->indice])) {
        e->codigo_error = 2;
        return 0;
    }

    while (isdigit((unsigned char)e->texto[e->indice])) {
        numero = numero * 10 + (e->texto[e->indice] - '0');
        e->indice++;
    }

    return negativo ? -numero : numero;
}

static long multiplicacion_division(Evaluador *e) {
    long resultado = numero_o_parentesis(e);

    while (e->codigo_error == 0) {
        char operador;

        saltar_espacios(e);
        operador = e->texto[e->indice];

        if (operador != '*' && operador != '/') {
            break;
        }

        e->indice++;

        long valor = numero_o_parentesis(e);

        if (e->codigo_error != 0) {
            return 0;
        }

        if (operador == '*') {
            resultado *= valor;
        } else {
            if (valor == 0) {
                e->codigo_error = 1;
                return 0;
            }

            resultado /= valor;
        }
    }

    return resultado;
}

static long expresion(Evaluador *e) {
    long resultado = multiplicacion_division(e);

    while (e->codigo_error == 0) {
        char operador;

        saltar_espacios(e);
        operador = e->texto[e->indice];

        if (operador != '+' && operador != '-') {
            break;
        }

        e->indice++;

        long valor = multiplicacion_division(e);

        if (e->codigo_error != 0) {
            return 0;
        }

        if (operador == '+') {
            resultado += valor;
        } else {
            resultado -= valor;
        }
    }

    return resultado;
}

long evaluar_expresion(const char *fuente, int *error) {
    Evaluador evaluador;
    long resultado;

    if (error == NULL) {
        return 0;
    }

    *error = 0;

    if (fuente == NULL) {
        *error = 2;
        return 0;
    }

    evaluador.texto = fuente;
    evaluador.indice = 0;
    evaluador.codigo_error = 0;

    resultado = expresion(&evaluador);

    saltar_espacios(&evaluador);

    if (evaluador.codigo_error == 0 &&
        evaluador.texto[evaluador.indice] != '\0') {
        evaluador.codigo_error = 2;
    }

    *error = evaluador.codigo_error;

    if (*error != 0) {
        return 0;
    }

    return resultado;
}
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion10_c.c
/* harness_sesion10_c.c - auditoria automatica de la Sesion 10
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion10_c.c artesano_sesion10_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion10.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 10 - Mini-evaluador de expresiones (Hito 1) ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int e=0; long r=evaluar_expresion("2+3*4",&e); (void)e; verificar_int("\"2+3*4\"", (long)(14), (long)(r)); }
    { int e=0; long r=evaluar_expresion("(2+3)*4",&e); (void)e; verificar_int("\"(2+3)*4\"", (long)(20), (long)(r)); }
    { int e=0; long r=evaluar_expresion("10/0",&e); (void)r; verificar_int("\"10/0\" -> error 1", (long)(1), (long)(e)); }
    { int e=0; long r=evaluar_expresion("2+",&e); (void)r; verificar_int("\"2+\" -> error 2", (long)(2), (long)(e)); }
    { int e=0; long r=evaluar_expresion("  7 ",&e); (void)e; verificar_int("\"  7 \"", (long)(7), (long)(r)); }
    { int e=0; long r=evaluar_expresion("-5+10",&e); (void)e; verificar_int("\"-5+10\"", (long)(5), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion10.md" "\n");
    return 0;
}
````
